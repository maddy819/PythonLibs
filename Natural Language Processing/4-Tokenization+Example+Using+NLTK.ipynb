```python
##  Tokenization
## Sentence-->paragraphs
import nltk
from nltk.tokenize import sent_tokenize
from nltk.tokenize import word_tokenize
from nltk.tokenize import wordpunct_tokenize
from nltk.tokenize import TreebankWordTokenizer
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
# A Paragraph / Document of text data is called as Corpus
corpus="""Hello Welcome,to Krish Naik's NLP Tutorials.
Please do watch the entire course! to become expert in NLP.
"""

print(corpus)
```

```python
# Sentence Tokenization
documents=sent_tokenize(corpus)

print(f"type(documents): {type(documents)} \n")
print(f"documents: {documents} \n")

for sentence in documents:
    print(f"sentence: {sentence} \n")
```

```python
# Word Tokenization
print("Word Tokenization\n")
print(word_tokenize(corpus))
```

```python
# Word Tokenization for each sentence
print("Word Tokenization for each sentence\n")
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
# WordPunct Tokenization
# It splits all punctuations into separate tokens
print("WordPunct Tokenization\n")
print(wordpunct_tokenize(corpus))
```

```python
# TreebankWordTokenizer
# It is similar to WordPunct but it does not split 's and n't
tokenizer=TreebankWordTokenizer()

print("TreebankWordTokenizer\n")
print(tokenizer.tokenize(corpus))
```
