```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Example with predefined data
x = [1, 2, 3, 4, 5]
y = [2, 11, 5, 7, 6]

plt.plot(x, y, marker='o')
plt.title("Line Chart")
plt.show()
```

```python
# Additional example with random data generated using numpy
x = np.arange(25)
y = np.random.randint(1, 50, size=25)

plt.plot(x, y, marker='s')
plt.title("Line Chart")
plt.show()
```

```python
# Another example with multiple points of y axis for same x axis value using numpy

# Set random seed for reproducibility
# This is important for generating the same random numbers across different runs
x = np.arange(1,6)
y = np.random.randint(1, 500, size=(5, 2)) # 5 rows (x values), 2 columns (y values)
# Shuffle rows
rows = np.random.permutation(y.shape[0])

# Shuffle columns
cols = np.random.permutation(y.shape[1])

shuffled = y[rows][:, cols]

print("x:", x)
print("y:\n", shuffled)

plt.figure(figsize=(6,4))
for col in range(shuffled.shape[1]):
    print(f"Plotting column {col} with values {shuffled[:,col]}")
    plt.plot(x, shuffled[:,col], marker='o', label=f'Column {col}')  # marker only

#plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Multiple y values per x (2D array)')
plt.legend()
plt.show()
```

```python
# Example to show sales per dayof week
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
sales = [150, 200, 300, 250, 400, 500, 450]

plt.plot(days, sales, marker='o')
plt.title("Sales per Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Sales")
plt.grid()
plt.show()
```
