```python
import numpy as np
import pandas as pd
from scipy import interpolate
import matplotlib.pyplot as plt
```

```python
# Example with predefined data
x = [1, 2, 3, 4, 5]
y = [2, 11, 5, 7, 6]

plt.plot(x, y, marker='o')
plt.title("Line Chart")
plt.show()
```

```python
# Additional example with random data generated using numpy
x = np.arange(25)
y = np.random.randint(1, 50, size=25)

plt.plot(x, y, marker='s')
plt.title("Line Chart")
plt.show()
```

```python
# Another example with multiple points of y axis for same x axis value using numpy

# Set random seed for reproducibility
# This is important for generating the same random numbers across different runs
x = np.arange(1,6)
y = np.random.randint(1, 500, size=(5, 2)) # 5 rows (x values), 2 columns (y values)
# Shuffle rows
rows = np.random.permutation(y.shape[0])

# Shuffle columns
cols = np.random.permutation(y.shape[1])

shuffled = y[rows][:, cols]

print("x:", x)
print("y:\n", shuffled)

plt.figure(figsize=(6,4))
for col in range(shuffled.shape[1]):
    print(f"Plotting column {col} with values {shuffled[:,col]}")
    plt.plot(x, shuffled[:,col], marker='o', label=f'Column {col}')  # marker only

#plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Multiple y values per x (2D array)')
plt.legend()
plt.grid()
plt.xlim(0, 6)
plt.ylim(0, 600)
plt.show()
```

```python
# Example to show sales per dayof week
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
sales = [150, 200, 300, 250, 400, 500, 450]

plt.plot(days, sales, marker='o')
plt.title("Sales per Day of Week")
plt.xlabel("Day of Week")
plt.ylabel("Sales")
plt.grid()
plt.ylim(0, 600)
plt.show()
```

```python
# Sample DataFrame
employees_df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "salary": [30000, 50000, 55000, 40000, 70000],
    "performance_score": [3, 6, 7, 3.5, 6.5]
})

# Convert names to numeric positions
positions = np.arange(len(employees_df))

# Plot using positions, but keep string labels
plt.plot(positions, employees_df["salary"], marker="o", label='Employees Salary')
plt.plot(positions, employees_df["performance_score"] * 10000, marker="s", label='Performance Score')
plt.xticks(positions, employees_df["name"])  # show employee names instead of numbers
plt.xlabel("Employee")
plt.ylabel("Salary")
plt.title("Employee Salaries")
plt.grid()
plt.xlim(-1, len(employees_df))  # add some padding on x-axis
plt.ylim(0, 80000)
plt.legend()
plt.show()
```

```python
# Known data
x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 1, 4, 9, 16])  # y = x^2

# New x values (for interpolation)
x_new = np.linspace(0, 4, 50)
print("x_new: \n", x_new)

# Linear interpolation
f_linear = interpolate.interp1d(x, y, kind='linear')
y_linear = f_linear(x_new)
print("y_linear: \n", y_linear)

# Cubic spline interpolation
f_cubic = interpolate.interp1d(x, y, kind='cubic')
y_cubic = f_cubic(x_new)
print("y_cubic: \n", y_cubic)

# Plot
plt.scatter(x, y, color='red', label="Data points")
plt.plot(x_new, y_linear, label="Linear Interpolation")
plt.plot(x_new, y_cubic, label="Cubic Spline Interpolation")
plt.legend()
plt.grid()
plt.show()
```
